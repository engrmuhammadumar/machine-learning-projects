```python
import scipy.io
import os
import numpy as np

# Define the path to your .mat files
folder_path = r'E:\CP Dataset\[20200428] PUMP DATA\VIbration_ALL'  # Update this with your directory

# List of files for each class (example based on your file names)
file_names = [
    r'E:\CP Dataset\[20200428] PUMP DATA\VIbration_ALL\impeller_crack_vibration_ALL.mat',
    r'E:\CP Dataset\[20200428] PUMP DATA\VIbration_ALL\mechanical_seal_hole_vibration_ALL.mat',
    r'E:\CP Dataset\[20200428] PUMP DATA\VIbration_ALL\mechanical_seal_scratch_vibration_ALL.mat',
    r'E:\CP Dataset\[20200428] PUMP DATA\VIbration_ALL\normal_vibration_ALL.mat'
]

# Dictionary to hold the loaded data
data_dict = {}

# Load each .mat file
for file_name in file_names:
    file_path = os.path.join(folder_path, file_name)
    
    # Load the .mat file
    mat_data = scipy.io.loadmat(file_path)
    
    # Access the 'concatenated_rows' variable
    data_dict[file_name] = mat_data['concatenated_rows']

# Now you have your data loaded for all classes in data_dict
# You can access each dataset like: data_dict[file_names[0]] for the first file
```

```python
import matplotlib.pyplot as plt
from scipy.signal import welch

# Parameters for Welch
fs = 25600  # Sampling frequency (adjust based on your data)
window_length = 1024
overlap_length = 512
fft_points = 2048

# Function to apply Welch's algorithm to a signal
def apply_welch(signal, fs, window_length, overlap_length, fft_points):
    f, pxx = welch(signal, fs, window='blackman', nperseg=window_length, noverlap=overlap_length, nfft=fft_points)
    return f, pxx

# Loop through each class and each signal in the dataset
for class_name, class_data in data_dict.items():
    print(f"Processing {class_name}")
    
    # Process each signal in this class
    for i in range(class_data.shape[0]):  # Assuming each row is a signal
        signal = class_data[i, :]  # Extract the ith signal
        
        # Apply Welch algorithm
        f, pxx = apply_welch(signal, fs, window_length, overlap_length, fft_points)
        
        # Plot the Power Spectral Density
        plt.figure()
        plt.semilogy(f, pxx)
        plt.title(f'Welch PSD - {class_name} - Signal {i+1}')
        plt.xlabel('Frequency (Hz)')
        plt.ylabel('Power/Frequency (dB/Hz)')
        plt.show()
```
